```python
"""
Aidan Repchik
Assignment#1
"""

gym_member = "Alex Alliton" #String
preferred_weight_kg = 20.5  #Float
highest_reps = 25           #Int
membership_active = True    #Boolean

# Dictionary containing friends' names and minutes spent on workouts
workout_stats = {
    "Tom": (30, 35, 40),
    "David": (40, 45, 50),
    "Dan": (40, 50, 60)
}
total_minutes_stats = {}

for i in workout_stats:
    minutes = workout_stats[i]
    total_minutes = minutes[0] + minutes[1] + minutes[2]
    total_minutes_stats[i + "_Total"] = total_minutes

workout_stats.update(total_minutes_stats)

# 2 Dimensional list containing workout minutes for each friend and activity
workout_list = [
    list(workout_stats["Tom"]),
    list(workout_stats["David"]),
    list(workout_stats["Dan"])
]
print(workout_list)

# yr_minutes is yoga and running time in minutes for all friends
yr_minutes = [row [:2] for row in workout_list]
print("Yoga and Running For for All Friends:")
print(yr_minutes)

# wl_minutes is the minutes for weightlifting for the last two friends
wl_minutes = [row[2] for row in workout_list[-2:]]
print("Weightlifting Minutes for the Last Two Friends:")
print(wl_minutes)

for i in total_minutes_stats:
    if total_minutes_stats[i] >= 120:
        friend_name = i.replace("_Total", "")
        print(f"Great job staying active, {friend_name}!")

controller = input("Would you like to search for a Member (y/n) ? ")
if controller == "y":
    Name = input("Enter Name: ")
    if Name in workout_stats:
        if isinstance(workout_stats[Name], tuple):
            print(f"Workout minutes for {Name}:")
            total_minutes = 0
            activities = ["Yoga", "Running", "Weightlifting"]
            for i, minutes in enumerate(workout_stats[Name]):
                print(f"{activities[i]}: {minutes} minutes")
                total_minutes += minutes
            print(f"Total workout minutes: {total_minutes}")
        else:
            print(f"Friend {Name} not found in the records.")
    else:
        print(f"Friend {Name} not found in the records.")
elif controller == "n":
    print("Ok!")
else:
    print("Please enter either 'y' or 'n'")


max_minutes = max(total_minutes_stats.values())
min_minutes = min(total_minutes_stats.values())

max_friend = [friend.replace("_Total", "") for friend, minutes in total_minutes_stats.items() if minutes == max_minutes][0]
min_friend = [friend.replace("_Total", "") for friend, minutes in total_minutes_stats.items() if minutes == min_minutes][0]

print(f"The friend with the highest total workout minutes is {max_friend} with {max_minutes} minutes.")
print(f"The friend with the lowest total workout minutes is {min_friend} with {min_minutes} minutes.")
```
